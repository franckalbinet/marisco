# Install marisco on Windows with Anaconda
> How to set up a marisco development environment on Windows

marisco is developed in Jupyter notebooks using nbdev. The notebooks live in `nbs/`, and the generated package in `marisco/`. This guide covers a Windows setup with Anaconda. If you prefer a lighter environment manager, use uv instead (see the uv section below).

## Prerequisites

- Anaconda Navigator installed.
- A GitHub account (needed for the editable install from the repository).

## Create an environment

1. Open Anaconda Navigator, select `Environments`.
2. Select `Create`.
3. Name your environment, for example `mariscoDev`.
4. Select a Python version. marisco currently supports Python 3.9 and above.

## Install the package

In Anaconda Navigator, go to `Home`, select your environment, and install the `CMD.exe` Prompt. Launch it and make sure your environment name appears in brackets.

To install the released package:

```
pip install marisco
```

To install an editable version for development, download the marisco repository from [GitHub](https://github.com/franckalbinet/marisco), or clone it:

1. Unzip or clone the repository.
2. In the `CMD.exe` Prompt, change to the package directory.
3. Install the package in editable mode:

```
cd path_to_downloaded_package
pip install -e .
```

## The uv route

[uv](https://docs.astral.sh/uv/) is a fast Python package and environment manager. It replaces the conda/mamba environment workflow with a single tool.

Install uv, then create a virtual environment and install marisco:

```
uv venv mariscoDev
uv pip install -e .
```

The first command creates a `.venv` in the current directory. The second installs the package into it. Activate the environment before running the command line tools:

```
source .venv/bin/activate
```

## Run the command line tools

marisco ships two command line tools: `marisco-ingest` to curate a dataset into a NetCDF file, and `marisco-export` to export the NetCDF contents to CSV. Both are available after installation.

For a first check, list the available handlers with an invalid name:

```
marisco-ingest not-a-handler
```

The error message lists all valid handler names.

## Launch Jupyter

Install Jupyter Notebook in Anaconda and launch it. The development notebooks are in `nbs/`.

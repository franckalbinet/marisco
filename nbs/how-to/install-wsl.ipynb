# Install marisco with VS Code and the Windows Subsystem for Linux

> How to set up a marisco development environment on Windows using WSL and VS Code

marisco is developed in Jupyter notebooks using nbdev. The notebooks live in `nbs/`, and the generated package in `marisco/`. This guide covers a Windows setup with WSL and VS Code.

## Prerequisites

- Visual Studio Code.
- Windows Terminal.
- A GitHub account.

## Install WSL

1. Download and install [Visual Studio Code for Windows](https://code.visualstudio.com/download).
2. Download and install [Windows Terminal](https://www.microsoft.com/store/productId/9N0DX20HK701).
3. In Windows Terminal, open a Windows PowerShell terminal and install WSL:

```
wsl --install
```

To see the available Ubuntu versions:

```
wsl --list --online
```

To install a specific version:

```
wsl --install -d Ubuntu
```

Complete the installation by creating a user (for example `marisco`) and a password.

To list the installed Linux distributions:

```
wsl --list --verbose
```

## Connect VS Code to WSL

1. Restart VS Code and install the `Remote Development` extension (Ctrl+Shift+X).
2. Restart VS Code, go to Remote Explorer, and confirm that WSL Targets are available.
3. Select `Connect in current window` for your Ubuntu environment.
4. The current environment is shown in the bottom left of VS Code.

## Set up the environment

1. In VS Code, open a new terminal (menu bar, `Terminal`, `New Terminal`).
2. Navigate to your home directory:

```
cd /home/{username}
```

### With mamba or conda

Install [Mambaforge](https://github.com/conda-forge/miniforge), which provides both `conda` and `mamba`. Then create an environment:

```
mamba create -n mariscoDev python=3.12
mamba activate mariscoDev
```

### With uv

[uv](https://docs.astral.sh/uv/) is a fast Python package and environment manager. It replaces the conda/mamba workflow with a single tool:

```
uv venv mariscoDev
source .venv/bin/activate
```

## Install the package

To install the released package:

```
pip install marisco
```

To install an editable version for development, clone the repository and install:

```
git clone https://github.com/franckalbinet/marisco.git
cd marisco
pip install -e .
```

## Run the command line tools

marisco ships two command line tools: `marisco-ingest` to curate a dataset into a NetCDF file, and `marisco-export` to export the NetCDF contents to CSV. Both are available after installation.

For a first check, list the available handlers with an invalid name:

```
marisco-ingest not-a-handler
```

The error message lists all valid handler names.

## Launch Jupyter

With VS Code, install the Jupyter extension and open `nbs/index.ipynb`. The development notebooks are in `nbs/`.

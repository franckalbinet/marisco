# Cheatsheet
> The commands you meet when you work on marisco with the agent

[Stage 2](find-your-way.ipynb) explains what each tool does. In the commands below, replace `<...>` with your own value.

## Terminal

| Command | What it does |
|---|---|
| `pwd` | Print the current folder |
| `ls`, `ls -la` | List the current folder. `-la` adds hidden files, sizes and dates |
| `cd <folder>`, `cd ..`, `cd` | Go into a folder, up one level, or back to your home folder. `cd ~` also goes to your home folder |
| `mkdir <folder>` | Create a folder |
| `cat <file>`, `less <file>` | Print a file, or page through it (`q` quits) |
| `cp <from> <to>`, `mv <from> <to>` | Copy, or move or rename |
| `rm <file>` | Delete a file. There is no trash bin and no undo |
| `which <command>` | Print where the shell finds a program |
| Ctrl+C | Stop the running command |
| `clear`, Ctrl+L | Clear the screen |
| Up arrow, Down arrow | Step back and forward through the commands you typed before. Press Enter to run one again |
| Tab | Complete the name of a file, folder or command. Press Tab twice to list the possible completions |

## Git

| Command | What it does |
|---|---|
| `git status` | Show the current branch, and the changed and staged files |
| `git diff`, `git diff --staged` | Show changes not yet staged, or staged changes |
| `git log --oneline -10` | Show the last 10 commits |
| `git switch main` | Go to the `main` branch |
| `git switch -c <branch>` | Create a branch and go to it |
| `git add <file>` | Stage a change for the next commit |
| `git commit -m "<message>"` | Record the staged changes as a commit |
| `git push -u origin <branch>` | Send a new branch to GitHub |
| `git push` | Send new commits on the current branch to GitHub |
| `git pull` | Bring in new commits from GitHub |
| `git branch -d <branch>` | Delete a branch that is already merged |
| `git stash`, `git stash pop` | Put uncommitted changes aside, and bring them back |

These commands discard work. Approve them only when you know what they discard:

| Command | What it discards |
|---|---|
| `git restore <file>` | Uncommitted changes to the file |
| `git reset --hard` | All uncommitted changes, in every file |
| `git clean -fd` | Files that Git does not track |
| `git branch -D <branch>` | A branch, even if it is not merged |
| `git push --force` | Commits on GitHub that your folder does not have |

## GitHub CLI

| Command | What it does |
|---|---|
| `gh issue list` | List open issues |
| `gh issue view <number>` | Show an issue and its comments |
| `gh issue create` | Open a new issue |
| `gh pr create` | Open a PR for the current branch |
| `gh pr list` | List open PRs |
| `gh pr view --web` | Open the current branch's PR in the browser |
| `gh pr checks` | Show the CI results of the current branch's PR |
| `gh pr checkout <number>` | Get the branch of a PR, to try it out |

## Workspace

| Command | What it does |
|---|---|
| `source ~/aai-ws/.venv/bin/activate` | Activate the workspace environment in this terminal |
| `ws-sync` | Pull every repository in the workspace and update the Python packages |
| `ws-status` | Show uncommitted changes in every repository of the workspace |
| `which python` | Show which Python environment is active |

## nbdev

| Command | What it does |
|---|---|
| `nbdev-export` | Regenerate the package in `marisco/` from the notebooks |
| `nbdev-clean` | Remove Jupyter metadata from the notebooks |
| `nbdev-test` | Run the tests of every notebook |
| `nbdev-test <notebook>` | Run the tests of one notebook |
| `nbdev-readme` | Regenerate `README.md` from `nbs/index.ipynb` |

## marisco

| Command | What it does |
|---|---|
| `marisco-ingest <handler> <output>` | Encode a provider dataset as a MARIS NetCDF file |
| `marisco-ingest <handler> <output> --src <file>` | The same, for handlers that read a local file |
| `marisco-ingest not-a-handler` | List the handlers |
| `marisco-export <file.nc>` | Write one CSV file for each sample type in a NetCDF file |
| `marisco-poll` | List the Zenodo records that have a newer version than the one a handler reads, and the TITANICA datasets that no handler reads |

## Claude Code

| Command | What it does |
|---|---|
| `claude` | Start the agent in the current folder |
| Esc | Stop what the agent is doing |
| Shift+Tab | Switch permission mode: manual, accept edits, plan, auto. The status bar shows the current mode |
| `/clear` | Start a new conversation |
| `/help` | List the commands |
| `/exit` | Quit |

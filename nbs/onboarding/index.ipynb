# Onboarding
> A path for MARIS team members who will create marisco handlers with the help of an AI agent

These pages take you from a new computer to your first handler. You work with Claude Code, an AI agent that reads the marisco code, proposes changes and runs commands for you. You don't need to write Python from memory. You need to understand enough to judge what the agent proposes, to approve or refuse each action, and to recover when something goes wrong.

The path has four stages. Each stage ends with checks that show you are ready for the next one.

1. [Set up your machine](setup.ipynb): install the terminal tools, Git, uv, Claude Code and the aai-ws workspace, then add marisco to it.
2. [Find your way](find-your-way.ipynb): what the terminal, Git, GitHub, uv and nbdev do in marisco, and your first pull request. The [cheatsheet](cheatsheet.ipynb) lists the commands.
3. [Supervise the agent](supervise.ipynb): what to check before you approve an action, when to stop the agent, how to review and undo its work, and when to move to auto mode.
4. Write your first handler: a small real dataset, following [Writing a new handler](../how-to/writing-a-handler.ipynb), reviewed with the maintainer. Not written yet.
